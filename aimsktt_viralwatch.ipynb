# Loading `training_table_final.csv`

```python
import pandas as pd

df_training_table_final = pd.read_csv('output/training_table_final.csv')
print('First 5 rows of df_training_table_final:')
display(df_training_table_final.head())
print('\nInformation about df_training_table_final:')
df_training_table_final.info()
```

```python
print('\nDescription of df_training_table_final:')
display(df_training_table_final.describe())
```

### Visualizations for `df_training_table_final`

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure 'date' column is in datetime format for plotting
df_training_table_final['date'] = pd.to_datetime(df_training_table_final['date'])

# Sort by date for correct time series plotting
df_training_table_final = df_training_table_final.sort_values(by='date')
```

#### Distribution of Key Numerical Features (Histograms)

```python
numerical_cols = [
    'cumulative_confirmed_cases',
    'cumulative_suspected_cases',
    'cumulative_suspected_deaths',
    'new_confirmed_cases',
    'new_suspected_cases',
    'min_minutes_to_active_zone',
    'pop_count',
    'pop_density',
    'rt_proxy'
]

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(3, 3, i + 1)
    sns.histplot(df_training_table_final[col], kde=True, bins=15)
    plt.title(f'Distribution of {col}')
    plt.xlabel('')
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

#### Time Series Trends for Cases

```python
plt.figure(figsize=(14, 6))
sns.lineplot(data=df_training_table_final, x='date', y='cumulative_confirmed_cases', marker='o')
plt.title('Cumulative Confirmed Cases Over Time')
plt.xlabel('Date')
plt.ylabel('Cumulative Confirmed Cases')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(14, 6))
sns.lineplot(data=df_training_table_final, x='date', y='new_confirmed_cases', marker='o', color='orange')
plt.title('New Confirmed Cases Over Time')
plt.xlabel('Date')
plt.ylabel('New Confirmed Cases')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Description and Visualizations for `df_insp_training_window`

```python
print('Description of df_insp_training_window:')
display(df_insp_training_window.describe())

# Ensure 'date' column is in datetime format and sort for plotting
df_insp_training_window['date'] = pd.to_datetime(df_insp_training_window['date'])
df_insp_training_window = df_insp_training_window.sort_values(by='date')

# Time Series Trends for Cases in df_insp_training_window
plt.figure(figsize=(14, 6))
sns.lineplot(data=df_insp_training_window, x='date', y='cumulative_confirmed_cases', marker='o')
plt.title('Cumulative Confirmed Cases Over Time (Training Window)')
plt.xlabel('Date')
plt.ylabel('Cumulative Confirmed Cases')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(14, 6))
sns.lineplot(data=df_insp_training_window, x='date', y='new_confirmed_cases', marker='o', color='orange')
plt.title('New Confirmed Cases Over Time (Training Window)')
plt.xlabel('Date')
plt.ylabel('New Confirmed Cases')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
